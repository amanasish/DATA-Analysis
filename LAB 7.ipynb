# 1) Simple Linear Regression

```python
import pandas as pd
data=pd.read_csv("dataset_regression.csv")
data.head()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
# feature and target selection since SLR so only one feature
X = data[['Size']]
y = data['Price']
```

```python
model = LinearRegression() 
```

```python
#train
model.fit(X, y)
```

```python
print("Simple Linear Regression Equation:")
print("Intercept (b0):", model.intercept_)
print("Coefficient (b1 for Size):", model.coef_[0])
```

```python
# prediction
y_pred = model.predict(X)
```

```python
# r sq. & sse ( sum of sq error ) evaluation
r2 = r2_score(y, y_pred)
```

```python
sse = ((y - y_pred) ** 2).sum()
```

```python
print("R2 Score:", r2)
print("SSE:", sse)
```

Ans: R2 since 0.9153232065490018 which is closer to 1 so better model ~ 91.53% and strong relationship between size and price

as SSE is more we look only R2 

```python
import matplotlib.pyplot as plt

plt.scatter(X, y)
plt.plot(X, y_pred, color='red')
plt.xlabel("Size")
plt.ylabel("Price")
plt.title("Simple Linear Regression")
plt.show()
```
